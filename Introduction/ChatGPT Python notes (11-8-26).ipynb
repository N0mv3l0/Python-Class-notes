## while loops

continues while a condition is true

```python
number= 3

while number <= 10:
    print(number)
    number+= 1
```

```python
number= 0

while number < 5:
    print(number)
    number 
    +1      #classic infinite while loop
```

```python
number= 0

while number < 5:
    print(number)
    number = number +1  
```

```python
number= 0

while number < 5:
    print(number)
    number +=1  
```

So whenever you write a while loop, ask yourself:

"What is going to make my condition become False?"
