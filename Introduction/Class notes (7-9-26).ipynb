# Pandas data structures

- Series: 1 dimensional label array
- Can hold any data type
    - int (integer/ whole number/ without a decimal)
    - str (has a combination of characters- letters and numbers and symbols). It is always shown with ""/''
    - float (number with a decimal)
- Has an index for each value. It tells you the position of the value, see example below.

```python
import pandas as pd

s= pd.Series([10,15,25,30])

display(s)
```
