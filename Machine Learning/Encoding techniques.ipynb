#Technique 1: Label encoding

Label Encoding: assigns a unique integer to each category.
It is the simpletest possible mapping from symbolic categories to numeric values. It  preserves uniqueness while minimizing dimensionality. It answers the question: "How can I convert categories into numbers without expanding the feature space?"

Simple definition: 
Lable encoding is a data preprocessing technique used to convert categorical (text) data into numbers so that the machine learing model can understand it.

# Why do we use label encoding?
ML models often work with numbers rather than text.

Label encoding helps us to:

a. Convert text into numbers

b. prepare data for machine learning

c. represent categories in a format that models can process.

# Technique 2: One-Hot Encoding

One-hot encoding creates a separate column for each category.

Let's use our colour example again.

Colour	Blue	Green	Red

Red	     0       0		1

Blue	 1	     0	    0

Green	 0	     1	    0

Red	     0	     0	    1

Each row gets a 1 in the column matching its category and 0 everywhere else.

This avoids suggesting that one colour is greater than another.

```python
#Example 1:

#Call in pandas library since you are creating a sample dataframe
# Import the pandas library to work with tabular data

import pandas as pd

#Create a sample data frame

df = pd.DataFrame({
    "Colour": ["Red", "Blue", "Green", "Red"]

})
```

```python
#show the dataframe

print(df)
```

```python
#We want to encode the colors using the "get_dummies" method cause we are encoding straight from the pandas dataframe.
# Apply One-Hot Encoding to the Colour column
# Each unique colour will become its own column
# dtype=int ensures the encoded values are 0 and 1

encoded_df = pd.get_dummies(
    df,                         #original column/data
    columns=["Colour"],         #column we want to encode
    dtype=int                   #data type = integer cause we dont want any decimals
) 
```

```python
# Display the encoded DataFrame
print(encoded_df)
```

#What's happening?
- pd.get_dummies() identifies the unique colours.
- It creates one column for each colour.
- A value of 1 indicates the colour in that row.
- A value of 0 indicates that the colour does not match that row.
For example, the first row has Colour_Red = 1 because the original colour is Red.

```python
#Same dataset but this time using mehod 2 (one hot encoding using sklearn library)

# Import pandas to create and display the DataFrame
import pandas as pd

# Import OneHotEncoder from scikit-learn
from sklearn.preprocessing import OneHotEncoder

# Create a DataFrame containing colours
df = pd.DataFrame({
    "Colour": ["Red", "Blue", "Green", "Red"]
})
```

```python
print(df)
```

```python
# Create the OneHotEncoder object
# sparse_output=False returns a regular array instead of a sparse matrix
# handle_unknown="ignore" allows unseen colours when transforming new data
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore",
    dtype=int
)
```

```python
# Learn the colour categories and encode the Colour column
encoded_array = encoder.fit_transform(df[["Colour"]])

# Convert the encoded array into a DataFrame
# get_feature_names_out() supplies the names of the new columns
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(["Colour"]),
    index=df.index
)
```

```python
# Display the encoded DataFrame
print(encoded_df)
```

#Let's understand the new sklearn terms

Code	Basic explanation

OneHotEncoder()	                Creates the encoder.

fit_transform()	                Learns the categories and encodes them in one step.

sparse_output=False	            Returns a regular array that is easier to view.

get_feature_names_out()	        Retrieves the names for the encoded columns.

handle_unknown="ignore"	        Handles categories not seen when the encoder was fitted.

#What does fit_transform() mean?

Think of it as two actions:
- fit: Learn which colours exist in the training data.
- transform: Convert those colours into numerical columns.

In Machine Learning projects, you will often use fit_transform() on training data, then use transform() on test data so the same encoding is applied consistently.

Note: sparse_output=False works in recent scikit-learn versions. Older versions use sparse=False instead.

#Technique 3: Ordinal Encoding


This is one of the most important techniques to understand because it is similar to label encoding, but there is a key difference.

Ordinal encoding is used when categories have a meaningful order or ranking.

```python
from sklearn.preprocessing import OrdinalEncoder

# 1. Define the input data (3 samples, 1 feature/column)
education = [["Primary"], ["Secondary"], ["Tertiary"]]

# 2. Initialize the OrdinalEncoder
# 'categories' must be a list containing one inner list per feature.
# Since we only have 1 feature, we pass a list with 1 inner list containing all 3 order categories.
encoder = OrdinalEncoder(
    categories=[["Primary", "Secondary", "Tertiary"]]
)

# 3. Fit the encoder to the data and transform the text into integers
encoded = encoder.fit_transform(education)

# 4. Print the result
# Output will be: [[0.], [1.], [2.]]
print(encoded)
```

```python
#Example 2

# Step 1: Import pandas to create and work with tables (DataFrames)
import pandas as pd

# Step 2: Import OrdinalEncoder from the scikit-learn library
# OrdinalEncoder converts categories into numbers based on a specific order
from sklearn.preprocessing import OrdinalEncoder

# Step 3: Create a DataFrame containing product quality ratings
df = pd.DataFrame({
    "Quality": ["High", "Low", "Medium", "High", "Low"]
})


print(df)
```

```python
# Step 4: Create the OrdinalEncoder object
# We specify the correct order of the categories:
# Low = 0, Medium = 1, High = 2 (here we are telling python the correct order of ranking- very important)
# This is important because quality has a meaningful ranking
encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

```

```python

# Step 5: Fit the encoder to the Quality column and transform the text into numbers
# fit_transform() learns the specified categories and encodes each value
# Double square brackets [[ ]] select the column as a 2D DataFrame,
# which is the format OrdinalEncoder expects
df["Quality_encoded"] = encoder.fit_transform(
    df[["Quality"]]
)

```

```python

# Step 6: Display the original column alongside the encoded column
print(df)
```

#Technique 4: Target Encoding


This one is slightly more advanced, but let's make it easy.

Target encoding replaces each category with a statistic calculated from the outcome we want to predict. A common statistic is the average target value for that category.

Imagine you're trying to predict whether a customer will buy a product. Your target column is:
- 1 = Bought
- 0 = Did not buy

```python
import pandas as pd 

# 1. Create a sample dataset
# "City" is our categorical feature, and "Bought" is our binary target (1 = Yes, 0 = No)

df = pd.DataFrame({
    "City": [
        "Durban", "Durban", "Durban",
        "Cape Town", "Cape Town",
        "Johannesburg", "Johannesburg"
    ],
    "Bought": [1, 1, 0, 1, 0, 0, 0]
})

print(df)
```

```python
# 2. Calculate the mean of the target variable ('Bought') for each unique category ('City')
# This calculates the success rate (probability of buying) per city

city_rates = df.groupby("City")["Bought"].mean()

print(city_rates)
```

```python
# 3. Map those calculated mean rates back into the original dataframe
# This replaces each city name with its corresponding target probability score
df["City_encoded"] = df["City"].map(city_rates)
```

```python
# 4. Display the encoding dictionary (the mapping rule)
print("--- Target Encoding Map (Mean per City) ---")
print(city_rates)
```

```python
# 5. Display the final dataframe with the new numerical column
print("\n--- Final Dataframe with Encoded Feature ---")
print(df)
```

#Quick Breakdown of What is Happening

• What it does: Target encoding replaces a text category (like "Durban") with a number that represents how often the target event (Bought = 1) happens for that specific category.

• The Math:

	• Durban has three entries: [1, 1, 0]. The mean is \(2 / 3 = \mathbf{0.667}\)

	• Cape Town has two entries: [1, 0]. The mean is \(1 / 2 = \mathbf{0.500}\)
    
	• Johannesburg has two entries: [0, 0]. The mean is \(0 / 2 = \mathbf{0.000}\)



```python
#Example 2

import pandas as pd

# 1. Create a sample real estate dataset
# "Neighborhood" is our categorical feature.
# "Price_Millions" is our continuous target variable (the value we want to predict).


df = pd.DataFrame({
    "Neighborhood": [
        "Greenpoint", "Greenpoint", "Morningside",
        "Morningside", "Morningside",
        "Umhlanga", "Umhlanga"
    ],
    "Price_Millions": [4.5, 5.5, 2.0, 2.5, 3.0, 6.0, 8.0]
})

print(df)
```

```python
# 2. Calculate the average (mean) house price for each neighborhood
# This groups the rows by neighborhood and extracts the mathematical average price.

Neighborhood_rates = df.groupby("Neighborhood")["Price_Millions"].mean()

print(Neighborhood_rates)
```

```python
# 3. Map those calculated mean rates back into the original dataframe
# This replaces each city name with its corresponding target probability score
df["Neighborhood_encoded"] = df["Neighborhood"].map(Neighborhood_rates)
```

```python
# 4. Display the encoding dictionary (the mapping rule)
print("--- Target Encoding Map (Mean per Neighborhood) ---")
print(Neighborhood_rates)
```

```python
# 5. Display the final dataframe with the new numerical column
print("\n--- Final Dataframe with Encoded Feature ---")
print(df)
```

# When should you use it?
Target encoding can be helpful when a categorical column has many unique values, such as hundreds of cities, suppliers or product categories. It can represent useful relationships without creating hundreds of separate columns.

Important warning: data leakage!

If we calculate target encoding using information from the same records we're trying to train and evaluate the model on, the model can indirectly see information about the answers. This can make its performance look better than it really is.

In practice, target encoding should be performed carefully within the training process, often using cross-validation or out-of-fold encoding, with smoothing for categories that have very few records.

# Technique 5: Frequency Encoding

Instead of looking at whether customers bought something, frequency encoding looks at how often each category appears in the dataset.

```python
# Step 1: Import pandas to work with data in tables (DataFrames)
import pandas as pd

# Step 2: Create a DataFrame containing product categories
df = pd.DataFrame({
    "Category": [
        "Shoes", "Clothing", "Shoes",
        "Electronics", "Shoes", "Clothing"
    ]
})

display(df)
```

```python

# Step 3: Count how often each category appears in the dataset
# normalize=True converts the counts into proportions (frequencies)
frequency_map = df["Category"].value_counts(normalize=True)

print(frequency_map)
```

```python

# Step 4: Replace each category with its calculated frequency
# .map() matches each category to its frequency in frequency_map
# For example, Shoes appears 3 times out of 6 rows, so its frequency is 0.5
df["Category_encoded"] = df["Category"].map(frequency_map)

```

```python

# Step 5: Display the original categories alongside their encoded values
print(df)
```

Three important things to remember
- value_counts(normalize=True) calculates the proportion of the dataset represented by each category.
- .map(frequency_map) replaces each text category with its corresponding frequency.
- Categories that appear more often receive higher frequency values, but that does not mean they are more profitable or more important.

#When should you use it?
- When a column has many unique categories.
- When how common or rare a category is might help the model.
- When you want to avoid creating a large number of extra columns.
However, two different categories can have the same frequency. Also, a frequent category is not automatically more important or more profitable. Frequency only tells us how common it is.
