#basic line graph 

```python
import matplotlib.pyplot as plt

#1 create a sample data for a line graph
x=[1,2,3,4,5]
y=[2,4,3,5,7]

#2 create a line plot
plt.plot(x,y, marker= "o")

#Add labels and titles
plt.title("My first graph")
plt.xlabel("my X")
plt.ylabel("my Y")

#Adding data labels
for i in range(len(x)):
    plt.text(x[i], y[i],f"({x[i]},{y[i]})")

#Display the graph
plt.show
```

```python
import matplotlib.pyplot as plt

#1 create a sample data for a line graph
x=[1,2,3,4]
y=[10,20,25,30]

#2 create a line plot
plt.figure(figsize=(12,4))
plt.plot(x,y, marker= "o", linestyle= '--', color= "black", markersize=10) #markercolor= "blue")
plt.grid(True)
plt.legend()

#Add labels and titles
plt.title("Monthly Sales")
plt.xlabel("months")
plt.ylabel("sales")



#Display the graph
plt.show
```

```python
import matplotlib.pyplot as plt

#Step 1: create sample data
fruits=["Apples","Bananas","Cherries","Kiwi"]
quantity= [10,15,7,12]

#Step 2: create a bar chart 
plt.bar(fruits,
        quantity)

#Step 3: Add labels
plt.title("Fruit Quantities")
plt.xlabel("fruits")
plt.ylabel("quantity")

#Step 4: show the graph
plt.show()
```

```python
import matplotlib.pyplot as plt

#Step 1: create sample data
fruits=["Apples","Bananas","Cherries","Kiwi"]
quantity= [10,15,7,12]

#Step 2: create a pie chart
plt.pie(quantity,
        labels=fruits)

plt.title("Fruits distribution")

#display pie chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#Step 1: create sample data
months=["Jan","Feb","Mar","Apr","May"]
revenue= [100,200,100,300,400]

#Step 2: create a bar chart
plt.bar(months,  #categorucal data - x axis
        revenue,  #numeric data - y axis
        color= "blue",
        edgecolor= "black"
        )

plt.title("Monthly revenue")
plt.xlabel("months")
plt.ylabel("revenue")
plt.legend()
plt.grid(True)

#display bar chart
plt.show()
```
